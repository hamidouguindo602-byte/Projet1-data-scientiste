Avant de traiter les vaaleurs manquantes on va  partage partage les données en deux données l'un numerique et l'autre categorique

```python
data_num=[]
data_cat=[]
for c,i in enumerate(data.dtypes):
    if i==object:
        data_cat.append(data.iloc[:,c])
    else:
        data_num.append(data.iloc[:,c])
```

```python
data_cat
```

Transformer cette liste des liste en dataframe 

```python
data_cat=pd.DataFrame(data_cat)
data_num=pd.DataFrame(data_num)
data_num
```

On vas transposé pour avoir le dataset qu'on souhaite

```python
data_cat=data_cat.transpose()
data_num=data_num.transpose()
data_cat
```

Pour les valeurs categorire on va remplacer par des valeurs qui se repete le plus

```python
data_cat=data_cat.apply(lambda x:x.fillna(x.value_counts().index[0]))
data_cat.isnull().sum().sort_values(ascending=False)
```

Pour les valeurs numerique on va remplacer par des valeurs qui leurs precede

```python
data_num.fillna(method='bfill',inplace=True)
data_num.isnull().sum().sort_values(ascending=False)
```

On va remplacer les y et n des la colonne de resultat par 1 =y et 0 =n

```python
target=data_cat['Loan_Status']
data_cat=data_cat.drop(['Loan_Status'],axis=1)
target=target.map({'Y':1,'N':0})
target
```

```python
data_cat.describe(include='O')
```

```python
data_cat['Dependents'].unique()
```

```python
data_cat['Gender']=data_cat['Gender'].map({'Male':1,'Female':0})
data_cat['Married']=data_cat['Married'].map({'Yes':1,'No':0})
data_cat['Self_Employed']=data_cat['Self_Employed'].map({'Yes':1,'No':0})
data_cat['Education']=data_cat['Education'].map({'Graduate':1,'Not Graduate':0})
data_cat['Property_Area']=data_cat['Property_Area'].map({'Rural':0,'Urban':1,'Semiurban':2})
data_cat['Dependents']=data_cat['Dependents'].map({'0':0,'1':1,'2':2,'3+':3})
```

```python
from sklearn.model_selection import StratifiedShuffleSplit
```

```python
sss=StratifiedShuffleSplit(n_splits=1,test_size=0.2,random_state=42)
for train,test in sss.split(x,y):
    x_train,x_test=x.iloc[train],x.iloc[test]
    y_train,y_test=y.iloc[train],y.iloc[test]
print("x_train taille",x_train.shape)
print("x_test taille",x_test.shape)
print("y_train taille",y_train.shape)
print("y_test taille",y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
model=LogisticRegression(random_state=42)
```

Fonction de prediction

```python
def accu(y_true,y_pre,retu=False):
    acc=accuracy_score(y_true,y_pre)
    if retu:
        return acc
    else:
        print("la precition est:",acc)
def train_test_eval(model,x_train,x_test,y_test,y_train):
    model.fit(x_train,y_train)
    accu(y_test,model.predict(x_test))
train_test_eval(model,x_train,x_test,y_test,y_train)
```
